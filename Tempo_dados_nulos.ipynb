```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/drive/MyDrive/python/weatherAUS.csv')
```

```python
df
```

```python
df.dtypes
```

```python
df_nulos = df.groupby('Location').apply(lambda x: x.isnull().sum())
```

```python
df_nulos
```

```python
df.isnull().sum()
```

```python
df_quantitativas = df.select_dtypes(include=['int64','float64'])
df_quantitativas
```

```python
df_qualitativas = df.select_dtypes(include=['object'])
df_qualitativas
```

```python
df_qualitativas.isnull().sum()
```

```python
df_quantitativas.isnull().sum()
```

```python
df[df_quantitativas.columns] = df[df_quantitativas.columns].fillna(df[df_quantitativas.columns].median())
```

```python
df.isnull().sum()
```

```python
moda_rain_tomorrow = df.RainTomorrow.mode().tolist()
moda_rain_today = df.RainToday.mode().tolist()
moda_wind_gust_dir = df.WindGustDir.mode().tolist()
moda_wind_dir_9am = df.WindDir9am.mode().tolist()
moda_wind_dir_3pm = df.WindDir3pm.mode().tolist()
```

```python
print(moda_rain_tomorrow)
print(moda_rain_today)
print(moda_wind_gust_dir)
print(moda_wind_dir_9am)
print(moda_wind_dir_3pm)
```

```python
df[df_qualitativas.columns] = df[df_qualitativas.columns].fillna(df[df_qualitativas.columns].mode().iloc[0])
```

```python
df.isnull().sum()
```

```python
df = df.drop('Date', axis=1)
```

```python
df.to_csv('dataset__chuva_tratado.csv', index=False)

from google.colab import files
files.download('dataset__chuva_tratado.csv')
```
