```python
def quanti_outliers(df, col):
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)
  IQR = Q3 - Q1

  return df[(df[col] < Q1 - 1.5*IQR) | (df[col] > Q3 + 1.5*IQR)]



```

```python
outliers_rain = quanti_outliers(df, 'Rainfall')
outliers_rain
```

```python
outliers_Evaporation = quanti_outliers(df, 'Evaporation')
outliers_Evaporation
```
